```python
import statistics as st
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = [1,4,5,8,9,10,10,11,12,13]
```

```python
numeric_data = data
x = st.mean(numeric_data)
print("Mean is",x)
mean = x
x = st.median(numeric_data)
print("Median is",x)
median = x
x = st.mode(numeric_data)
print("Mode is",x)
mode = x
x = st.stdev(numeric_data)
print("Standard Deviation is",x)
std_dev = x
x = st.variance(numeric_data)
print("Variance is",x)
variance = x
```

```python
labels = ['Mean','Median','Mode','Standard Deviation','Variance']
values = [mean,median,mode,std_dev,variance]
```

```python
plt.figure(figsize=(8,5))
plt.bar(labels,values,color='black')
plt.title("Summary Statistics")
plt.ylabel("Value")
plt.grid(axis='both')
plt.tight_layout()
plt.savefig("summary_statistics_plot.png")
plt.show
```
