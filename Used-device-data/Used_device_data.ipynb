NOW WE WILL START PLOTTING GRAPH

plt.tight_layout()
Automatically adjusts subplot spacing to avoid overlap.

```python
import matplotlib.pyplot as plt
import seaborn as sns
fig = plt.figure(figsize=(10,10))
for i in range(len(num_cols)):
    plt.subplot(4,3,i+1)
    sns.boxplot(data=data_no_missing,x=data_no_missing[num_cols[i]])
plt.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize=(10,10))

for i in range(len(num_cols)):
    plt.subplot(4,3,i+1)
    sns.kdeplot(data=data_no_missing,x=data_no_missing[num_cols[i]])

plt.tight_layout()
plt.show()
```
